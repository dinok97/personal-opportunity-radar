```python
import numpy as np
from sentence_transformers import SentenceTransformer

MODEL = "jinaai/jina-embeddings-v5-text-nano"   # same name + revision you use in the app
st = SentenceTransformer(MODEL, trust_remote_code=True)

# load the same jobs (adjust to your DB access)
jobs = retrieve_jobs("ml jobs with masters in computer science", top_n_limit=20)
```

```python
texts = [j.job.embedding_text for j in jobs]
```

```python
texts
```

```python
doc_vecs = st.encode(texts, task="retrieval", prompt_name="document",
                     normalize_embeddings=True, batch_size=16, show_progress_bar=True)
```

```python
def raw_search(query: str):
    qv = st.encode(query, task="retrieval", prompt_name="query", normalize_embeddings=True)
    sims = doc_vecs @ qv                      # scores from the full embedding_text
    for i in np.argsort(-sims):
        print(f"{sims[i]:.3f}  doc #{i}")
```

```python
for q in [".NET C# developer"]:
    print(f"\nQUERY: {q}")
    raw_search(q)
```

```python

TESTS = [
    ("computer vision",                          [10, 7, 4]),
    ("medical imaging deep learning",            [4]),
    ("MLOps platform and infrastructure",        [5]),
    ("RAG and LLM applications",                 [18, 0, 8, 12, 6]),
    ("speech recognition and voice assistants",  [19]),
    ("PhD in climate and machine learning",      [2]),
    ("credit risk modelling in banking",         [17]),
    ("AI regulation and compliance",             [13]),
    ("project manager for AI projects",          [1]),
    ("AI research for wireless communication",   [11]),
]
```

```python
def evaluate(name, ranker):
    hits = 0
    for q, expected in TESTS:
        order = ranker(q)
        top3 = order[:3]
        ok = any(e in top3 for e in expected)
        hits += ok
        print(f"  {'✓' if ok else '✗'} {q:42s} top3={top3}  expected={expected}")
    print(f"  => {name}: {hits}/{len(TESTS)}\n")
```
